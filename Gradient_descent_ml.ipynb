```python
from google.colab import files
import io
import pandas as pd

# 1. Prompt to upload file
uploaded = files.upload()

# 2. Get the filename of the uploaded file
file_name = list(uploaded.keys())[0]
print(f'Uploaded file: {file_name}')

# 3. Read the file (e.g., if it's a CSV, read into a Pandas DataFrame)
df = pd.read_csv(io.BytesIO(uploaded[file_name]))
df.head()
```

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(df[['age', 'affordibility']], df.bought_insurance,
                                                    test_size=0.2, random_state=25)
```

```python
len(x_train)
```

```python
df.shape
```

```python
x_train_scaled = x_train.copy()
x_train_scaled['age'] = x_train_scaled['age'] / 100

x_test_scaled = x_test.copy()
x_test_scaled['age'] = x_test_scaled['age'] / 100
```

```python
model = keras.Sequential([
    keras.layers.Dense(1, input_shape=(2,), activation='sigmoid', kernel_initializer='ones',
                       bias_initializer='zeros')
])
model.compile(optimizer='adam',
              loss='binary_crossentropy', #log loss functions
              metrics=['accuracy'])
model.fit(x_train_scaled, y_train, epochs=500)
```

```python
model.evaluate(x_test_scaled, y_test)
```
