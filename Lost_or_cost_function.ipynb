```python
import numpy as np

y_predicted = np.array([1,1,0,0,1])
y_true = np.array([0.3,0.7,1,0,0.5])
```

```python
def mae(y_true, y_predicted):
  total_error = 0
  for yt, yp in zip(y_true, y_predicted):
    total_error += abs(yt - yp)
  print("Total error: ", total_error)

  mae = total_error / len(y_true)
  print("MAE:", mae)
  return mae
    # print(yt, yp)
  #   total_error += abs(yt - yp)
  # print("Total error: ", total_error)
  # mae = total_error / len(y_true)
  # print("MAE: ", mae)
  # return mae
```

Log Loss Function

```python
# Clip y_predicted to be between epsilon and 1 - epsilon
y_predicted_clipped = np.clip(y_predicted, eplison, 1 - eplison)

# Compute the binary cross entropy loss
-np.mean(y_true * np.log(y_predicted_clipped) + (1 - y_true) * np.log(1 - y_predicted_clipped))
```

```python
def log_loss(y_true, y_predicted):
  y_predicted_new = [max(float(i), eplison) for i in y_predicted]
  y_predicted_new = [min(float(i), 1-eplison) for i in y_predicted]
  y_predicted_new = np.array(y_predicted_new)
  return -np.mean(y_true * np.log(y_predicted_new) + (1 - y_true) * np.log(1 - y_predicted_new))
```

```python
log_loss(y_true, y_predicted)
```
