```python
import math

def sigmoid(x):
  """
  Computes the sigmoid activation function.
  Maps any real-valued number into a range between 0 and 1.
  """
  return 1 / (1 + math.exp(-x))

def tanh(x):
  """
  Computes the hyperbolic tangent (tanh) activation function.
  Maps any real-valued number into a range between -1 and 1.
  """
  return (math.exp(x) - math.exp(-x)) / (math.exp(x) + math.exp(-x))

def relu(x):
  """
  Computes the Rectified Linear Unit (ReLU) activation function.
  Returns 0 if the input is negative, and the input itself if it is positive.
  """
  return max(0, x)
```

```python
### Explanation of Activation Functions

These functions are widely used in neural networks as **activation functions** to introduce non-linearity into the model:

1. **Sigmoid:** Maps input values to an S-shaped curve between `0` and `1`. Often used in output layers for binary classification tasks.
2. **Tanh (Hyperbolic Tangent):** Maps input values to an S-shaped curve between `-1` and `1`. It is zero-centered, which often makes optimization easier compared to sigmoid.
3. **ReLU (Rectified Linear Unit):** Simply passes positive values through and replaces negative values with zero. It is highly popular due to its computational efficiency and helping reduce the vanishing gradient problem.
```
